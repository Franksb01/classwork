```python
# Chess with lists

pawns = ["p"]*8
cool = ["R", "N", "B", "Q", "K", "B", "N", "R"]
empty = [" "]*8

chess=[
    cool.copy(),
    pawns.copy(),
    empty.copy(),
    empty.copy(),
    empty.copy(),
    empty.copy(),
    pawns.copy(),
    cool.copy()
]
chess
```

```python
chess[0][1]=" "
chess[2][2]="N"
chess
```

```python
# tuple

#not mutable(no changing),
# allows duplicate
#indexed
tp1=(1,2,2,3, "something")
tp2=tuple((1,2,3,4,5))
print(tp1)
print(tp2)
```

```python
tp1.count(2)
tp1.index(2)
```

```python
#set
#can change(mutable)
# no duplicates
# unordered

set1= (1,2,1,2,1,2,3)# the bad way
set2= set([1,2,3,3,3,4,4,4,4])# the good way
```

```python
set2.remove(3)
len(set2)
```

```python
set2
```

```python
set2.difference(set1)
```
