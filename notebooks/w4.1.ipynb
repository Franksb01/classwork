```python
painting1 = {
    "color": "red",
    "size": "8x11"
}
if painting1["color"] == "blue":
   painting1["color"] = "red"
elif painting1["color"] == "black":
   painting1["color"] = "blue"
elif painting1["color"] == "red":
    painting1["color"] = "black"
else:
    print("no change to color")
    
print (painting1)
```

```python
#hastmaps

color_options = {
    "red": "black",
    "black": "blue",
    "blue": "black",
}

painting1["color"] = color_options[painting1["color"]]
painting1
```

```python
#loops
#while loop

counter = 0
while counter < 10:
    print(f"Counter = {counter}")
    counter += 1 # counter = counter + 1

is_true = True
while is_true:
    print(is_true)
    is_true = False
```
