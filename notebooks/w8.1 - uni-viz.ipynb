```python
import matplotlib.pyplot as plt
import seaborn as sns

import pandas as pd
from scipy import stats
url="https://raw.githubusercontent.com/nkmwicz/data-for-students/refs/heads/main/soccer-players.csv"

df = pd.read_csv(url)
df.head()
```

```python
plt.figure(figsize=(3, 2))
plt.xlabel('Day')
plt.ylabel('Sales ($)')
plt.title('Fig 1:Daily Coffee Sales')
plt.xlim(0, 7)
plt.ylim(0, 60)
plt.grid(True)
txt = "sales Items"
plt.text(0.20, 
         0.7, 
         txt, 
         ha="center", 
         color="red",
         transform=plt.gca().transAxes)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
bins = df['height_in_cm'].count() **(1/3)
plt.hist(df['height_in_cm'], bins=10,edgecolor='black')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
hist = sns.histplot(data=df, x="height_in_cm", kde=True)
plt.ylabel("Frequency")
hist.set(xlabel="Height (cm)")
mean_val = df['height_in_cm'].mean()
median_val = df['height_in_cm'].median()
plt.axvline(mean_val, 
            color='red', 
            linestyle='dashed',
            linewidth=1,
            label= 'mean')
plt.axvline(median_val,
            color = "yellow",
            linestyle='--',
            label = 'median'
            )

skew = stats.skew(df['height_in_cm'], nan_policy='omit')
kurt = stats.kurtosis(df['height_in_cm'], nan_policy='omit')
txt = f"""skew: {skew:.2f}
kurt: {kurt:.2f}
"""
plt.text(0.05,
         0.6,
         txt,
         transform=plt.gca().transAxes,
)
hist.set(title="histogram Soccer Player Heights")
sns.set_theme(style="whitegrid")
plt.legend()
plt.show()

plt.show()
```
